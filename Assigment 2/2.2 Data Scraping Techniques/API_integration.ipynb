```python
import requests
import pandas as pd
```

```python
url = "https://api.coingecko.com/api/v3/coins/markets"
```

```python

parameters = {
    "vs_currency": "usd",
    "order": "market_cap_desc",
    "per_page": 10,
    "page": 1
}
```

```python
response = requests.get(
    url,
    params=parameters
)
```

```python
print("Status Code:", response.status_code)

```

```python
if response.status_code == 200:

    print("API data received successfully.")


    # -----------------------------------------------------
    # STEP 6: CONVERT JSON DATA INTO PYTHON DATA
    # -----------------------------------------------------

    data = response.json()


    # -----------------------------------------------------
    # STEP 7: DISPLAY NUMBER OF RECORDS
    # -----------------------------------------------------

    print("Number of records received:", len(data))


    # -----------------------------------------------------
    # STEP 8: CONVERT DATA INTO DATAFRAME
    # -----------------------------------------------------

    df = pd.DataFrame(data)


    # -----------------------------------------------------
    # STEP 9: DISPLAY COMPLETE DATAFRAME
    # -----------------------------------------------------

    print("\nComplete DataFrame:")

    print(df)


    # -----------------------------------------------------
    # STEP 10: DISPLAY FIRST FEW ROWS
    # -----------------------------------------------------

    print("\nFirst few rows of the DataFrame:")

    print(df.head())


else:

    # -----------------------------------------------------
    # STEP 11: HANDLE UNSUCCESSFUL REQUEST
    # -----------------------------------------------------

    print("Unable to retrieve data from the API.")

    print("Status Code:", response.status_code)

```

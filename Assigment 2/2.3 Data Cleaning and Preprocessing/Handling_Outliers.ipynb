```python
import pandas as pd
import numpy as np
from scipy.stats import zscore
```

```python
# Create student data

data = {
    "Student": [
        "Amit",
        "Priya",
        "Rahul",
        "Sneha",
        "Vijay",
        "Neha",
        "Rohit",
        "Kiran",
        "Pooja",
        "Anita"
    ],

    "Marks": [
        85,
        78,
        92,
        88,
        95,
        82,
        90,
        87,
        89,
        300
    ]
}

# Create the DataFrame

df = pd.DataFrame(data)

# Display the original DataFrame

print("Original DataFrame:")
print(df)
```

```python
# Calculate the mean of the Marks column

mean_marks = df["Marks"].mean()

# Display the mean

print("Mean Marks =", mean_marks)
```

```python
# Calculate the standard deviation

std_marks = df["Marks"].std()

# Display the standard deviation

print("Standard Deviation =", std_marks)
```

```python
# Calculate the Z-score for every student's marks

df["Z_Score"] = zscore(df["Marks"])

# Display the DataFrame

print("DataFrame with Z-Scores:")
print(df)
```

```python
# Create a condition for identifying outliers

outlier_condition = df["Z_Score"].abs() > 2    # depending on this outliers will be identified. 

# Display only the rows that are outliers

print("Detected Outliers:")
print(df[outlier_condition])
```

```python
# Keep only rows that are NOT outliers

df_cleaned = df[~outlier_condition].copy()
# Remove the Z-score column because it was only
# required for detecting the outliers

df_cleaned = df_cleaned.drop(columns=["Z_Score"])

# Display the cleaned DataFrame

print("Cleaned DataFrame:")

print(df_cleaned)
```

```python
# Number of rows before removing outliers

rows_before = len(df)

# Number of rows after removing outliers

rows_after = len(df_cleaned)

# Calculate the number of removed outliers

outliers_removed = rows_before - rows_after


print("==============================================")
print("OUTLIER REMOVAL SUMMARY")
print("==============================================")

print("Rows before cleaning :", rows_before)
print("Rows after cleaning  :", rows_after)
print("Outliers removed     :", outliers_removed)
```

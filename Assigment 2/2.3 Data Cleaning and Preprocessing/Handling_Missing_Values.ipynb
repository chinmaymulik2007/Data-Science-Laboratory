```python
import pandas as pd
```

```python
# Create a dictionary containing student information

data = {
    "Name": ["Amit", "Priya", "Rahul", "Sneha", "Vijay"],
    "Age": [20, 21, None, 20, 22],
    "Marks": [85, None, 78, 92, None],
    "Attendance": [90, 88, 95, None, 85]
}

# Convert the dictionary into a pandas DataFrame

df = pd.DataFrame(data)

# Display the DataFrame

print("Original DataFrame:")
print(df)
```

```python
print("Number of Missing Values in Each Column:")
print(df.isnull().sum())
```

```python
# Create a copy of the original DataFrame

df_drop = df.copy()

# Remove rows containing missing values

df_drop = df_drop.dropna()

# Display the result

print("DataFrame After Dropping Rows with Missing Values:")
print(df_drop)
```

```python
# Create a copy of the original DataFrame

df_specific = df.copy()

# Replace all missing values with 0

df_specific = df_specific.fillna(0)

# Display the result

print("DataFrame After Filling Missing Values with 0:")
print(df_specific)
```

```python
# Create a copy of the original DataFrame

df_mean = df.copy()

# Calculate the mean of the Age column

age_mean = df_mean["Age"].mean()

# Replace missing Age values with the mean

df_mean["Age"] = df_mean["Age"].fillna(age_mean)


# Calculate the mean of the Marks column

marks_mean = df_mean["Marks"].mean()

# Replace missing Marks values with the mean

df_mean["Marks"] = df_mean["Marks"].fillna(marks_mean)


# Calculate the mean of the Attendance column

attendance_mean = df_mean["Attendance"].mean()

# Replace missing Attendance values with the mean

df_mean["Attendance"] = df_mean["Attendance"].fillna(attendance_mean)


# Display the calculated means

print("Mean Age:", age_mean)
print("Mean Marks:", marks_mean)
print("Mean Attendance:", attendance_mean)

print("\nDataFrame After Filling Missing Values with Column Mean:")
print(df_mean)
```

```python
print("==============================================")
print("ORIGINAL DATAFRAME")
print("==============================================")

print(df)


print("\n==============================================")
print("AFTER DROPPING MISSING VALUES")
print("==============================================")

print(df_drop)


print("\n==============================================")
print("AFTER FILLING WITH 0")
print("==============================================")

print(df_specific)


print("\n==============================================")
print("AFTER FILLING WITH COLUMN MEAN")
print("==============================================")

print(df_mean)
```

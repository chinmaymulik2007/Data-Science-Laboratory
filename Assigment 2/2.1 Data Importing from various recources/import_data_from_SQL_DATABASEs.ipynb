```python
import sqlite3
import pandas as pd
```

```python
connection=sqlite3.connect("student.db")
```

```python
cursor=connection.cursor()
```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS students(
    id INTEGER PRIMARY KEY,
    name TEXT,
    branch TEXT,
    marks INTEGER
)
""")
```

```python
cursor.execute("DELETE FROM students")
```

```python
students_data=[
    (1,"Akash","CSE",85),
    (2,"manas","CSE",95),
    (3,"dinesh","IT",50),
    (4,"rohit","CSE",62),
    (5,"yesh","AIML",75)
]
```

```python
cursor.executemany(
    "INSERT OR IGNORE INTO students (id,name,branch,marks)VALUES(?,?,?,?)",
    students_data
)
```

```python
connection.commit()
```

```python
query="SELECT * FROM students"
```

```python
df=pd.read_sql_query(query,connection)
```

```python
print("Complete Dataframe:")
print(df)
```

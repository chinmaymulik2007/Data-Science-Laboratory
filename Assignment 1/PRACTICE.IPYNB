```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("cereal.csv")
```

```python
print(df.head())
```

```python
plt.plot(df["name"], df["calories"])

plt.xlabel("Cereal Name")
plt.ylabel("Calories")
plt.title("Calories in Cereals")

plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(15, 6))

plt.plot(df["name"], df["calories"],color='green')

plt.xlabel("Cereal Name")
plt.ylabel("Calories")
plt.title("Calories in Cereals")

plt.xticks(rotation=270)
plt.show()
```

```python
plt.scatter(df["calories"], df["rating"],color='red')

plt.xlabel("Calories")
plt.ylabel("Rating")
plt.title("Calories vs Rating")

plt.show()
```

```python
plt.hist(df["calories"], bins=10)

plt.xlabel("Calories")
plt.ylabel("Number of Cereals")
plt.title("Distribution of Calories")

plt.show()
```

```python
plt.hist(df["calories"], bins=10, rwidth=0.5)

plt.xlabel("Calories")
plt.ylabel("Number of Cereals")
plt.title("Distribution of Calories")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

n, bins, patches = plt.hist(df["calories"], bins=10, rwidth=0.8)

colors = ["red", "blue", "green", "orange", "purple",
          "pink", "yellow", "cyan", "brown", "magenta"]

for patch, color in zip(patches, colors):
    patch.set_facecolor(color)

plt.xlabel("Calories")
plt.ylabel("Number of Cereals")
plt.title("Distribution of Calories")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("cereal.csv")

plt.scatter(df["calories"], df["rating"])

plt.xlabel("Calories")
plt.ylabel("Rating")
plt.title("Calories vs Rating")

plt.show()
```

```python
top10 = df.nlargest(10, "rating")

plt.figure(figsize=(12, 6))

plt.bar(top10["name"], top10["rating"], color="blue")

plt.xlabel("Cereal Name")
plt.ylabel("Rating")
plt.title("Top 10 Cereals by Rating")

plt.xticks(rotation=90)

plt.show()
```

```python
top10 = df.nlargest(10, "rating")

plt.figure(figsize=(12, 6))

colors = ["magenta","pink","magenta","pink","magenta","pink","magenta","pink","magenta","pink"]

plt.bar(top10["name"], top10["rating"], color=colors)

plt.xlabel("Cereal Name")
plt.ylabel("Rating")
plt.title("Top 10 Cereals by Rating")

plt.xticks(rotation=90)

plt.show()
```

```python
type_count = df["type"].value_counts()

plt.figure(figsize=(7, 7))

plt.pie(type_count, labels=type_count.index, autopct="%1.1f%%", colors=['magenta','pink'])

plt.title("Distribution of Cereal Types")

plt.show()
```
